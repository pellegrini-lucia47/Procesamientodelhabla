###Punto 3: APLICO EMBEDDINGS

###ELIJO EN ESTE CASO SPACY Y DENTRO DE ESTE EL MODELO PRE-ENTRENADO es_core_news_md EN ESPAÑOL.

```python
import spacy
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Descargar el modelo pre-entrenado de spaCy en español si no está instalado
!python -m spacy download es_core_news_md --quiet

# Cargar el modelo pre-entrenado de spaCy en español
nlp = spacy.load("es_core_news_md")

# Dataset
dataset_estetica = {
    "hola": "¡Hola! ¿En qué puedo ayudarte?",
    "adiós": "Hasta luego. ¡Gracias por hablar conmigo!",
    "turno": "¿Para qué servicio quieres turno?",
    "servicios": "Nuestros servicios incluyen Depilación, Limpieza Facial y Lifting de Pestañas.",
    "depilación": "Te envío turnos disponibles para depilación.",
    "limpieza facial": "Te envío turnos disponibles para Faciales.",
    "lifting de pestañas": "Te envío turnos disponibles para Lifting.",
    "¿cuánto cuesta?": "Te envío lista de precios actualizada.",
    "¿cuánto sale el lifting?": "Te envío lista de precios actualizada.",
    "¿cuánto sale la limpieza facial?": "Te envío lista de precios actualizada.",
    "¿cuánto sale la depilación?": "Te envío lista de precios actualizada.",
    "qué horarios atienden": "Nuestros horarios de atención son de 9hs a 12hs y 14hs a 18hs.",
    "¿trabajan los sábados?": "Atendemos solamente sábados por la mañana.",
    "¿atienden a hombres?": "No atendemos a hombres, solamente mujeres.",
    "¿atienden a niñas de 13 años?": "Sí, con el acompañamiento y consentimiento de su madre.",
    "formas de pago": "Recibimos transferencias o efectivo.",
    "reservas": "Para reservar tu turno deberás abonar el 30% del servicio.",
    "llegaré un poco tarde": "Recordá que la tolerancia del turno es de 10 minutos, de lo contrario el turno queda sin efecto.",
    "chau": "Gracias por hablar conmigo. ¡Hasta luego!",
    "salir": "Nos vemos."
}

preguntas = list(dataset_estetica.keys())
respuestas = list(dataset_estetica.values())
print("Dataset cargado exitosamente con", len(preguntas), "elementos.")
```

###COMPARO LOS DOS CHATBOTS, TANTO CON TFIDF COMO CON EMBEDDINGS.

```python
#Definimos la función de TF-IDF
def responder_tfidf(consulta_usuario):
    consulta_vec = vectorizer.transform([consulta_usuario])
    similitudes = cosine_similarity(consulta_vec, tfidf_matrix).flatten()
    indice = similitudes.argmax()
    return respuestas[indice], similitudes[indice]

# Agregamos la función de Embeddings
docs_preguntas = [nlp(p) for p in preguntas]

def responder_embeddings(consulta_usuario):
    doc_usuario = nlp(consulta_usuario)
    similitudes = [doc_usuario.similarity(doc_p) for doc_p in docs_preguntas]
    indice = max(range(len(similitudes)), key=similitudes.__getitem__)
    return respuestas[indice], similitudes[indice]

# Bucle de prueba
preguntas_prueba = [
    "¿Cuánto sale el lifting?",
    "¿Abren el fin de semana?",
    "¿Cuáles son los medios de pago?",
    "Llego tarde, perdón",
    "Quisiera sacar turno"
]

print("=== EVALUACIÓN COMPARATIVA (TF-IDF vs spaCy Embeddings) ===\n")
for q in preguntas_prueba:
    resp_t, score_t = responder_tfidf(q)
    resp_e, score_e = responder_embeddings(q)

    print(f"Consulta: '{q}'")
    print(f"  -> TF-IDF:     [{resp_t}] (Similitud: {score_t:.2f})")
    print(f"  -> Embeddings: [{resp_e}] (Similitud: {score_e:.2f})")
    print("-" * 60)
```

Preguntas que no pudo responder bien: "¿Abren el fin de semana?" y "Quisiera sacar turno", pero no los 2 sino que Embeddings en mi caso respondió mal 2 preguntas y TFIDF la de "¿abren el fin de semana?" ya que no lo relaciona con sábado. Con esto llego a la conclusión de que en este caso para mi Estética, me conviene el TF-IDF, aunque si la clienta escribe mal o tiene algún error de ortografía la consulta la toma como inválida y responde: lo siento, no entiendo tu pregunta. Y noté que el modelo de Spacy no llegó a encontrar la similitud entre fin de semana y sábado, y que al ser frases muy cortas también cada palabras pesa mucho en el promedio de los vectores y a veces no hay suficiente información para que entienda o capture bien la pregunta e intención de la misma.
